# A late update can arrive last
Run from this folder with Python and SQLite. Arrival order must not replace source order.

```python
import sqlite3
from pathlib import Path
db = sqlite3.connect(':memory:')
db.executescript(Path('setup.sql').read_text())
db.executescript(Path('solution.sql').read_text())
db.execute('DELETE FROM changes')
db.execute("INSERT INTO changes VALUES ('O1', 2, 'U', 1000)")
db.executescript(Path('solution.sql').read_text())
assert db.execute('SELECT source_version, amount_cents FROM current_orders').fetchone() == (3, 2500)
print('Late update refused; version 3 kept.')
```
