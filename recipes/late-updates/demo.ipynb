# A late update can arrive last
Run from this folder with Python and DuckDB. Arrival order must not replace source order.

```python
import duckdb
from pathlib import Path

folder = (
    Path("recipes/late-updates") if Path("recipes/late-updates").is_dir() else Path(".")
)

db = duckdb.connect(":memory:")
db.execute((folder / "setup.sql").read_text())
db.execute((folder / "solution.sql").read_text())
db.execute("DELETE FROM changes")
db.execute("INSERT INTO changes VALUES ('O1', 2, 'U', 1000)")
db.execute((folder / "solution.sql").read_text())
assert db.execute(
    "SELECT source_version, amount_cents FROM current_orders"
).fetchone() == (3, 2500)
print("Late update refused; version 3 kept.")
```
